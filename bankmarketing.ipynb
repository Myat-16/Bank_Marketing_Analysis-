4. Among customers with a balance greater than 5000 euros, what is the most common job?

5. For customers who have a personal loan (loan is 'yes'), what is the maximum balance for each marital status?

##### Ave balance for customers whose job is 'management'

```python
ave_balc_management = df[df['job'] == 'management']['balance'].mean()
ave_balc_management
```

##### Ave age for all 'single' customers

```python
single_ave_age = df[df['marital'] == 'single']['age'].mean()
single_ave_age
```

##### customers who have a housing loan, what is the count of people in each education level

```python
housing_loan_count = df[df['housing'] == 'yes'].groupby('education')['housing'].count()
housing_loan_count.sort_values(ascending=False)
```

##### cust balance greater than 5000 euros, what is the most common job?

```python
balance_greater_5000 = df[df['balance'] > 5000].groupby('job')['balance'].count()
balance_greater_5000.sort_values(ascending=False)
```

##### maximum balance for each marital status with loan 'yes'

```python
max_balance_personal_loan = df[df['loan'] == 'yes'].groupby('marital')['balance'].max()
max_balance_personal_loan.sort_values(ascending=False)
```

##### Ave balance for divorced people for each education level

```python
df.columns
```

```python
ave_balance_divorced_by_edu_level = df[df['marital'] == 'divorced'].groupby('education')['balance'].mean()
ave_balance_divorced_by_edu_level.sort_values(ascending=False)
```

##### average balance for customers who have a housing loan but no personal loan

```python
ave_housing_loan_no_personal_loan = df[(df['housing'] == 'yes') & (df['loan'] == 'no')].groupby('marital')['balance'].mean()
ave_housing_loan_no_personal_loan.sort_values(ascending=False)
```

##### Ave balance for each job of cust aged >=50

```python
ave_50andmore_age_balance = df[df['age'] >= 50].groupby('job')['balance'].mean()
```

```python
ave_50andmore_age_balance.sort_values(ascending=False)
```

##### Ave age by marital status with housing & personal loan

```python
ave_age_by_marital_with_loan = df[(df['housing'] == 'yes') & (df['loan'] == 'yes')].groupby('marital')['age'].mean()
ave_age_by_marital_with_loan.sort_values(ascending=False)
```
